## Question 1.4: Data Grouping and Aggregation (15 marks)

```python
# Write the code here with comments for explanation
# Group the dataset by product category and calculate required aggregations
category_summary = df.groupby('product_category').agg(
    total_transactions=('transaction_id', 'count'),
    total_quantity_sold=('quantity', 'sum'),
    total_sales_amount=('total_amount', 'sum')
)

# Display the aggregated results
category_summary
```

```python
# Identify the product category with the highest total sales amount
highest_sales_category = category_summary['total_sales_amount'].idxmax()

highest_sales_category
```

Describe the outputs


Grouping the data by product category reveals noticeable differences in transaction frequency, quantity sold, and total sales value. The category with the highest total sales amount generates the greatest share of revenue, indicating stronger customer demand and higher profitability compared to other categories. This imbalance suggests that certain categories play a more critical role in overall business performance. These insights can inform strategic decisions such as prioritising marketing investment, improving inventory planning, and focusing supplier negotiations on the most profitable categories.

## Question 1.5: Data Visualisation (20 marks)

```python
# Write the code here with comments for explanation
import matplotlib.pyplot as plt
```

```python
# Calculate the total number of transactions for each customer region
region_transactions = df['customer_region'].value_counts()

# Create a bar chart
plt.figure(figsize=(8, 5))
region_transactions.plot(kind='bar')

# Add labels and title
plt.xlabel('Customer Region')
plt.ylabel('Number of Transactions')
plt.title('Total Number of Transactions by Customer Region')

# Display the chart
plt.show()
```

```python
# Calculate total sales amount for each product category
category_sales = df.groupby('product_category')['total_amount'].sum()

# Create a pie chart
plt.figure(figsize=(7, 7))
plt.pie(category_sales, labels=category_sales.index, autopct='%1.1f%%')

# Add title
plt.title('Proportion of Total Sales Amount by Product Category')

# Display the chart
plt.show()
```

```python
# Create a scatter plot to show the relationship between quantity and total amount
plt.figure(figsize=(8, 5))
plt.scatter(df['quantity'], df['total_amount'], alpha=0.5)

# Add labels and title
plt.xlabel('Quantity')
plt.ylabel('Total Amount')
plt.title('Relationship Between Quantity and Total Amount')

# Display the chart
plt.show()
```

Describe the outputs


The bar chart illustrates the total number of transactions across different customer regions. Some regions generate significantly more transactions than others, suggesting geographical differences in customer engagement and purchasing frequency. This insight can help businesses target high-performing regions more effectively.

The pie chart shows the proportion of total sales amount contributed by each product category. Categories with larger shares dominate overall revenue, indicating revenue concentration. This information is valuable for identifying key revenue drivers and potential risks associated with over-reliance on a single category.

The scatter plot reveals a positive relationship between quantity purchased and total transaction amount. As the number of items increases, the total amount spent generally rises, confirming a consistent pricing structure. This relationship also highlights the impact of bulk purchases on overall sales revenue.

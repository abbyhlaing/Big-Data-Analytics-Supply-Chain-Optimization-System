## Q3: Query

Write a code to retrieve all wards where attendance is below 87%. Store them in a variable named `low_attendance_wards`. Print the number of documents.

# Q8: Inserting images

There are 5 images in a folder named `images` next to this Jupyter Notebook. Insert them into your database, with OBJECTIDs: 4000, 5000, 6000, 7000, and 8000 respectively. For example, the first image will have: `OBJECTID: 4000`. Use the following structure to insert these images as documents into your database. Note that:

- in the `file_name`, you need to store EXACTLY the same name as the name of the file, WITHOUT its extension.
- in the `original_size`, you need to store the size of the image and include its unit. For example, it could be `38.8 KB`. You can refer to recordings of the sessions to see how to find the size of a file.
- in the  `data`, you need to store the converted version of the image (base64).


Note: Please make sure that stored base64 images are decodable to the original images.

```python
# DO NOT WRITE ANY CODE IN THIS CELL. PLEASE USE THE NEXT CELL FOR YOUR ANSWER.

'''
"OBJECTID": 4000
"file_type": "image/jpg"
"object_type": "vehicle"
"file_name": "The name of the file"
"original_size": "The size of the file in text, NOT numbers"
"data": "base64 string"
'''
```

```python
# BEGIN SOLUTION
import os
import base64

# Folder path
folder_path = "images"

# Starting OBJECTID
object_id = 4000

# Loop through images
for file in os.listdir(folder_path):
    file_path = os.path.join(folder_path, file)

    # Read image and convert to base64
    with open(file_path, "rb") as img:
        encoded_string = base64.b64encode(img.read()).decode('utf-8')

    # Get file size in KB
    size_kb = os.path.getsize(file_path) / 1024
    size_text = str(round(size_kb, 1)) + " KB"

    # Remove extension from file name
    file_name = os.path.splitext(file)[0]

    # Create document
    image_doc = {
        "OBJECTID": object_id,
        "file_type": "image/jpg",
        "object_type": "vehicle",
        "file_name": file_name,
        "original_size": size_text,
        "data": encoded_string
    }

    # Insert into DB
    db.insert(image_doc)

    object_id += 1000
# END SOLUTION
```
